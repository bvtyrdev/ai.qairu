```python
# 1 тапсырма
import math
import random
def hill_climbing_1(values, start):
    current_idx = start
    n = len(values)
    while True:
        best_idx = current_idx
        if current_idx > 0 and values[current_idx - 1] > values[best_idx]:
            best_idx = current_idx - 1
        if current_idx < n - 1 and values[current_idx + 1] > values[best_idx]:
            best_idx = current_idx + 1
        if best_idx == current_idx:
            break
        current_idx = best_idx
    return current_idx, values[current_idx]


values1 = [1, 3, 5, 8, 6, 4, 2]
idx1, val1 = hill_climbing_1(values1, start=4)
print(f"Соңғы индекс: {idx1}, Соңғы мән: {val1}\n")
```

```python
# 4 тапсырма
def random_restart_4(values, attempts=5):
    best_idx = -1
    best_val = -float("inf")
    n = len(values)

    for _ in range(attempts):
        start = random.randint(0, n - 1)

        # Ішкі Hill Climbing функциясы
        current_idx = start
        while True:
            best_idx_hc = current_idx
            if current_idx > 0 and values[current_idx - 1] > values[best_idx_hc]:
                best_idx_hc = current_idx - 1
            if (
                current_idx < n - 1
                and values[current_idx + 1] > values[best_idx_hc]
            ):
                best_idx_hc = current_idx + 1
            if best_idx_hc == current_idx:
                break
            current_idx = best_idx_hc

        val = values[current_idx]
        if val > best_val:
            best_val = val
            best_idx = current_idx

    return best_idx, best_val


values4 = [1, 4, 7, 5, 3, 6, 9, 8]
for exp in range(1, 4):
    idx, val = random_restart_4(values4, attempts=5)
    print(
        f"Эксперимент {exp}  Табылған ең жақсы мән: {val} (Индекс: {idx})"
    )
print()
```

```python
# 5 тапсырма
def anneal_5(values, start, initial_temp=100.0, cooling_rate=0.95):
    current_idx = start
    current_val = values[current_idx]
    temp = initial_temp
    n = len(values)

    while temp > 0.01:
        neighbors = []
        if current_idx > 0:
            neighbors.append(current_idx - 1)
        if current_idx < n - 1:
            neighbors.append(current_idx + 1)

        next_idx = random.choice(neighbors)
        next_val = values[next_idx]
        delta = next_val - current_val

        if delta > 0 or random.random() < math.exp(delta / temp):
            current_idx = next_idx
            current_val = next_val

        temp *= cooling_rate

    return current_idx, current_val


values5 = [1, 4, 7, 5, 3, 6, 9, 8]
for run in range(1, 6):
    idx, val = anneal_5(values5, start=0)
    print(f"Іске қосу {run}  Соңғы индекс: {idx}, Соңғы мән: {val}")
```
